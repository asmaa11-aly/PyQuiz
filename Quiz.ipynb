```python
def run_quiz(selected_questions):

    score = 0

    for number, question in enumerate(selected_questions, start=1):

        print("\n" + "=" * 40)
        print(f"Question {number}/{len(selected_questions)}")
        print("=" * 40)

        print(question["ques"])

        for i, option in enumerate(question["options"], start=1):
            print(f"{i}. {option}")

        while True:

            try:
                user_choice = int(input("Choose your answer: "))

                if 1 <= user_choice <= len(question["options"]):
                    break

                print("Please choose a valid option.")

            except ValueError:
                print("Please enter a number.")

        selected_answer = question["options"][user_choice - 1]

        if selected_answer == question["answer"]:
            print("Correct!")
            score += 1

        else:
            print("Incorrect!")

    print("\n" + "=" * 40)
    print(f"Your score: {score}/{len(selected_questions)}")
    print("=" * 40)
```

```python
def choose_topic(questions):

    topics = []

    for question in questions:
        if question["topic"] not in topics:
            topics.append(question["topic"])

    print("Choose a topic:")

    for i, topic in enumerate(topics, start=1):
        print(f"{i}. {topic}")

    while True:

        try:
            choice = int(input("Choose a topic number: "))

            if 1 <= choice <= len(topics):
                break

            print("Please choose a valid number.")

        except ValueError:
            print("Please enter a number.")

    selected_topic = topics[choice - 1]

    selected_questions = []

    for question in questions:
        if question["topic"] == selected_topic:
            selected_questions.append(question)

    return selected_questions
```

```python
selected_questions = choose_topic(questions)

run_quiz(selected_questions)
```
